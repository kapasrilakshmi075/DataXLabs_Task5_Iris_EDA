```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load Dataset
df = pd.read_csv('Iris.csv')

# Display first 5 rows
df.head()
```

```python
# Dataset Information
print("Shape of Dataset:", df.shape)

print("\nColumns:")
print(df.columns)

print("\nDataset Info:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nStatistical Summary:")
df.describe()
```

Observation:
- The dataset contains 150 rows and 6 columns.
- No missing values are present.
- The dataset contains measurements of three Iris flower species.


```python
plt.figure(figsize=(6,4))
sns.countplot(x='Species', data=df)

plt.title('Species Distribution')
plt.xlabel('Species')
plt.ylabel('Count')

plt.show()
```

Observation:
- The dataset contains three species:
  Iris-setosa, Iris-versicolor, and Iris-virginica.
- Each species has 50 observations.
- The dataset is balanced with equal representation of all species.


```python
df.hist(figsize=(10,8))
plt.tight_layout()
plt.show()
```

Observation:
- Sepal and petal measurements show different distributions.
- Petal Length and Petal Width have wider variation.
- Most flower measurements are concentrated around the mean values.


```python
plt.figure(figsize=(8,5))
sns.boxplot(x='Species', y='PetalLengthCm', data=df)

plt.title('Petal Length by Species')
plt.show()
```

Observation:
- Iris-setosa has the smallest petal length.
- Iris-virginica has the largest petal length.
- Petal length varies significantly across species.


```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    x='PetalLengthCm',
    y='PetalWidthCm',
    hue='Species',
    data=df
)

plt.title('Petal Length vs Petal Width')
plt.show()
```

Observation:
- Species form distinct clusters.
- Iris-setosa is clearly separated from the other species.
- Petal measurements are useful for species classification.


```python
sns.pairplot(
    df.drop('Id', axis=1),
    hue='Species'
)

plt.show()
```

Observation:
- Strong separation exists between species.
- Petal features provide better classification than sepal features.
- Iris-setosa is the most distinguishable species.


```python
plt.figure(figsize=(8,6))

sns.heatmap(
    df.drop(['Id','Species'], axis=1).corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title('Correlation Heatmap')
plt.show()
```

Observation:
- Petal Length and Petal Width have a strong positive correlation.
- Sepal Length is moderately correlated with petal measurements.
- Sepal Width has weaker correlations with other features.


Summary of Findings

1. The Iris dataset contains 150 observations and no missing values.
2. All three species are equally represented.
3. Petal measurements show greater variation than sepal measurements.
4. Iris-setosa is clearly distinguishable from other species.
5. Petal Length and Petal Width have a strong positive correlation.
6. Scatter plots and pair plots reveal clear species clustering.
7. Petal features are highly useful for species classification.
